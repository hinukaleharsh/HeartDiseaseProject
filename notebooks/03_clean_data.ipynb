```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/heart_disease_uci.csv")
df.shape
```

```python
df = df.drop(columns=["id"])
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.shape
```

```python
print("Zero cholesterol rows:", (df["chol"] == 0).sum())
print("Zero blood pressure rows:", (df["trestbps"] == 0).sum())

df["chol"] = df["chol"].replace(0, np.nan)
df["trestbps"] = df["trestbps"].replace(0, np.nan)
```

```python
(df.isnull().mean() * 100).round(1)
```

```python
df = df.drop(columns=["ca", "thal", "dataset"])
df.shape
```

```python
for col in ["fbs", "restecg", "exang", "slope"]:
    df[col] = df[col].fillna("Unknown").astype(str)
```

```python
df["target"] = (df["num"] > 0).astype(int)
df = df.drop(columns=["num"])
df["target"].value_counts()
```

```python
print(df.shape)
print(df.isnull().sum())
df.head()
```

```python
df.to_csv("../data/processed/heart_clean.csv", index=False)
```
